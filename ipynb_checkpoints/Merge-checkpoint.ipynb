<h2>Fix the Skills Seperation:</h2>

```python
import ast

def fix_skills(x):
    if isinstance(x, str):
        # convert "['a', 'b']" → ['a','b']
        try:
            x = ast.literal_eval(x)
        except:
            return []
    return x
combined_df["skills_list"] = combined_df["skills_list"].apply(fix_skills)
```

```python
def split_embedded_lists(lst):
    cleaned = []
    for item in lst:
        if isinstance(item, str) and "," in item:
            parts = [p.strip() for p in item.split(",")]
            cleaned.extend(parts)
        else:
            cleaned.append(item)
    return cleaned

combined_df["skills_list"] = combined_df["skills_list"].apply(split_embedded_lists)
```

```python
def clean_skills_list(skills):
    """
    Flatten nested lists and remove extra spaces, lowercasing all skills.
    """
    cleaned = []
    for s in skills:
        if isinstance(s, list):
            # Flatten inner lists
            for item in s:
                cleaned.append(item.strip().lower())
        elif isinstance(s, str):
            # Some entries may be a single string with commas
            # Split by comma if needed
            parts = s.split(',')
            for p in parts:
                cleaned.append(p.strip().lower())
    # Remove duplicates
    return list(set(cleaned))

# Apply function to skills_list
combined_df['skills_list'] = combined_df['skills_list'].apply(clean_skills_list)
combined_df
```
