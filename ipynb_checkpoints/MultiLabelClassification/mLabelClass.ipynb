```python
import pandas as pd
import joblib
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.preprocessing import MultiLabelBinarizer
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.multiclass import OneVsRestClassifier

# --- 1. Load Data ---
df = pd.read_csv('../../all-courses-hybrid-45-skills.csv')

# --- 2. Prepare Features (CRITICAL FIX) ---
# Combine title and skills to give the model more context
df['text_all'] = df['course_title'].fillna('') + ' ' + df["skills_text"].fillna("")
X = df['text_all']

# --- 3. Prepare Labels ---
# Assuming your skill columns start from a certain index or are known
# Based on your files, they are the 45 skills
skill_cols = [
    'javascript','technical product management','financial accounting',
    'application development','system security','digital marketing','microsoft excel',
    'regulations and compliance','market analysis','swift programming','people management',
    'network architecture','theoretical computer science','bayesian network','negotiation',
    'people analysis','search engine SEO','sap','machine learning software',
    'user experience','database application','angular','basic descriptive statistics',
    'jira','spatial analysis','interactive design','strategy and operations',
    'big data','python programming','media strategy & planning','financial analysis',
    'human learning','google app engine','writing','android development','software testing',
    'risk management','game theory','web development','supply chain and logistics',
    'user experience design','agile software development','internet of things','power bi'
]

# Ensure labels are binary
Y = df[skill_cols]

# --- 4. Build Improved Pipeline ---
text_clf = Pipeline([
    ('tfidf', TfidfVectorizer(
        stop_words='english', # Removes "the", "and", etc.
        max_features=10000, 
        ngram_range=(1, 2), 
        min_df=1              # DO NOT drop words that appear only once
    )),
    ('clf', OneVsRestClassifier(LogisticRegression(solver='sag'), n_jobs=-1))
])

# --- 5. Train ---
text_clf.fit(X, Y)
print("Model trained successfully!")

# --- 6. Save Model AND Labels (CRITICAL FIX) ---
# Saving the labels ensures they are always in the same order as the model predictions
joblib.dump(text_clf, 'MultiLabelModel.pkl')
joblib.dump(skill_cols, 'skill_labels.pkl') 
print("Model and Labels saved!")
```

```python

def test_course_skills(text, pipeline):
    """
    Predicts skills and their probabilities for a given course text.
    
    Args:
        text (str): The course description text.
        pipeline (Pipeline): The fitted scikit-learn Pipeline object.
        
    Returns:
        list: A sorted list of (skill_name, probability) tuples.
    """
    # 1. Access the fitted OneVsRestClassifier from the pipeline
    ovr_classifier = pipeline.named_steps['clf']
    
    # 2. Get the prediction and probabilities
    pred = pipeline.predict([text])[0]
    proba = pipeline.predict_proba([text])[0]
    
    # 3. The classes_ attribute of OneVsRestClassifier stores the order 
    #    of the labels, which matches the order of probabilities.
    #    For multi-label tasks, the classes_ will be 0, 1, 2... for your N labels.
    #    We must ensure the skill names align with the probability order.
    #    In your case, this order should match your skill_cols list.

    results = []
    # ovr_classifier.classes_ is an array [0, 1, 2, ..., 44] for 45 skills
    for i, label_index in enumerate(ovr_classifier.classes_):
        # label_index will be 0, 1, 2, ...
        
        # Check if the class was predicted (val == 1)
        if pred[i] == 1: 
            skill_name = skill_cols[label_index] # Match the index to the name
            probability = proba[i]
            results.append((skill_name, round(probability, 3)))
            
    # Return the results sorted by probability in descending order
    return sorted(results, key=lambda x: -x[1])


# Example Usage (replace your existing call):
test_course_skills("This course teaches python programming, data science, deep learning, and google app engine.", text_clf)
```

```python
import numpy as np
import shap

vectorizer = text_clf.named_steps["tfidf"]
clf = text_clf.named_steps["clf"]
X_train_vec = vectorizer.transform(X_train)

def explain_course_with_shap(course_text, max_skills=3, top_words=5):
    """
    Returns SHAP-based word importances for the top predicted skills for this course.
    """
    # 1) Get predictions & probabilities for this course
    sample_vec = vectorizer.transform([course_text])
    proba = clf.predict_proba(sample_vec)[0]  # shape: (n_skills,)
    pred = (proba >= 0.5).astype(int)        # threshold; adjust if you want

    # 2) Choose which skills to explain: predicted = 1, sorted by prob
    skill_probs = [
        (skill_cols[i], proba[i], i)
        for i in range(len(skill_cols))
        if pred[i] == 1
    ]
    if not skill_probs:
        # no skill strongly predicted → just take top skills by prob
        skill_probs = sorted(
            [(skill_cols[i], proba[i], i) for i in range(len(skill_cols))],
            key=lambda x: -x[1]
        )[:max_skills]
    else:
        skill_probs = sorted(skill_probs, key=lambda x: -x[1])[:max_skills]

    feature_names = vectorizer.get_feature_names_out()
    explanations = {}

    # 3) For each selected skill, compute SHAP values
    for skill_name, p, idx in skill_probs:
        model_for_skill = clf.estimators_[idx]

        explainer = shap.LinearExplainer(
            model_for_skill,
            X_train_vec,
            feature_perturbation="interventional"
        )
        shap_vals = explainer.shap_values(sample_vec)[0]  # 1D: (n_features,)

        # Get top contributing words
        # sort features by absolute SHAP value
        top_idx = np.argsort(-np.abs(shap_vals))[:top_words]
        word_contribs = [
            (feature_names[j], float(shap_vals[j]))
            for j in top_idx
        ]

        explanations[skill_name] = {
            "probability": float(p),
            "word_contributions": word_contribs,
        }

    return explanations
```

```python
import pandas as pd
import joblib
import re
from nltk.stem import PorterStemmer
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.multiclass import OneVsRestClassifier
from sklearn.metrics import classification_report, hamming_loss

# ==========================================
# 1. LOAD DATA & CLEANING
# ==========================================
df = pd.read_csv('../../all-courses-hybrid-45-skills.csv')

# Dynamic Skill Detection: 
# Find all columns after 'skills_text'—these are your 45 target skills.
# This prevents the "KeyError" if a skill name is slightly different.
start_idx = df.columns.get_loc("skills_text") + 1
skill_cols = df.columns[start_idx:].tolist()

# Preprocessing: Combine Title and Skills Text for better context (Fixes LIME 0s)
df['combined_text'] = (
    df['course_title'].fillna('') + " " + 
    df['skills_text'].fillna('')
).str.lower().str.strip()

X = df['combined_text']
Y = df[skill_cols]

# ==========================================
# 2. STEMMING & TOKENIZATION (Fixes Consistency)
# ==========================================
# This ensures "Marketing", "Market", and "Marketing-based" are treated as the root "market"
stemmer = PorterStemmer()

def robust_tokenizer(text):
    # Remove special characters and numbers
    words = re.sub(r"[^a-zA-Z]", " ", text).split()
    # Return stemmed versions of words
    return [stemmer.stem(w) for w in words]

# ==========================================
# 3. BUILD THE BRAIN (The Pipeline)
# ==========================================
model_pipeline = Pipeline([
    ('tfidf', TfidfVectorizer(
        tokenizer=robust_tokenizer, 
        stop_words='english',
        ngram_range=(1, 1),    # Focus on single words for higher consistency
        sublinear_tf=True,     # Smoothes word weights to prevent "over-importance"
        min_df=2               # Ignore words that appear only once (typos)
    )),
    ('clf', OneVsRestClassifier(
        LogisticRegression(solver='sag', max_iter=1000, C=1.0), 
        n_jobs=-1
    ))
])

# ==========================================
# 4. TRAIN AND EVALUATE
# ==========================================
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=42)

print("Training model... this may take a moment.")
model_pipeline.fit(X_train, Y_train)

# Predictions for verification
predictions = model_pipeline.predict(X_test)

print("\n--- MODEL VERIFICATION REPORT ---")
# Weighted F1 is the best metric for your 45-skill project
print(classification_report(Y_test, predictions, target_names=skill_cols, zero_division=0))

h_loss = hamming_loss(Y_test, predictions)
print(f"Hamming Loss (lower is better): {h_loss:.4f}")

# ==========================================
# 5. EXPORT FOR INTEGRATION
# ==========================================
joblib.dump(model_pipeline, 'MultiLabelModel.pkl')
joblib.dump(skill_cols, 'skill_labels.pkl') 

print("\n🚀 SUCCESS!")
print("1. 'MultiLabelModel.pkl' updated with Stemming and Robust TF-IDF.")
print("2. 'skill_labels.pkl' saved to ensure your interface knows the skill order.")
```
