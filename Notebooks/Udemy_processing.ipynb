```python
import re
from collections import Counter, defaultdict

#read the coursera skills
SKILLS_FILE = "coursera_skills_list.txt"

# tokens that are generic stopwords regardless of frequency (extend if needed)
GLOBAL_GENERIC_TOKENS = {
    "programming","development","management","analysis","design","operations",
    "software","systems","data","business","project","principles","theory",
    "applications","tools","methods","practice","basics","models","model",
    "general","techniques","statistics","statistical","planning","research",
    "training","learning","skills","engineering","visualization","visual",
    "web","cloud","process","product"
}
```

```python
# string cleaning helper functions:
def clean_text(text):
    """Lowercase, remove punctuation (keep +.#), collapse spaces."""
    text = str(text).lower()
    text = re.sub(r"[^a-z0-9+.# ]", " ", text)
    return re.sub(r"\s+", " ", text).strip()

def normalize_skill_name(s):
    """Normalize single skill entry: remove parenthesis notes and extra spaces."""
    s = str(s).strip()
    # remove surrounding quotes/braces if any
    s = re.sub(r'^[\[\{"\']+|[\]\},"\']+$', '', s).strip()
    # remove parenthetical clarifiers like "react (web framework)" -> "react"
    s = re.sub(r'\([^)]*\)', '', s)
    s = re.sub(r'\s+', ' ', s).strip()
    return s.lower()
```

```python
# load the skills and clean them:
with open(SKILLS_FILE, "r", encoding="utf-8") as f:
    raw_skills = [line.strip() for line in f if line.strip()]

# normalize and deduplicate
skill_list = []
orig_map = {}   # map normalized -> original (for readability)
for s in raw_skills:
    norm = normalize_skill_name(s)
    if norm and norm not in skill_list:
        skill_list.append(norm)
        orig_map[norm] = s

print(f"Loaded {len(skill_list)} unique normalized skills.")
```

```python
# TOKEN FREQUENCIES (across skill list)
token_counts = Counter()
skill_tokens_map = {}
for skill in skill_list:
    toks = skill.split()
    skill_tokens_map[skill] = toks
    token_counts.update(toks)

# choose threshold to call a token 'generic' by frequency across skill list
# (tweak threshold depending on how large the skill list is; 6 is a good starting point)
GENERIC_TOKEN_FREQ_THRESHOLD = 6

# build generic token set = globally generic + frequent tokens
freq_generic_tokens = {tok for tok, cnt in token_counts.items() if cnt >= GENERIC_TOKEN_FREQ_THRESHOLD}
generic_tokens = set(GLOBAL_GENERIC_TOKENS) | freq_generic_tokens

# build core tokens (tokens considered informative)
core_tokens_map = {}
for skill, toks in skill_tokens_map.items():
    core = [t for t in toks if t not in generic_tokens]
    core_tokens_map[skill] = core

# Find skills that are problematic (no core tokens) -> we will drop or flag them
no_core_skills = [s for s, core in core_tokens_map.items() if len(core) == 0]
print(f"Skills with NO core tokens (will be flagged/dropped): {len(no_core_skills)}")
# print(no_core_skills[:50])

# Optional: drop or keep minimal list - here we remove them from skill_list to avoid spamming matches
safe_skill_list = [s for s in skill_list if s not in no_core_skills]
safe_skill_list = [s for s in safe_skill_list if s not in ["software as a service", "professional development"]]
print(f"Using {len(safe_skill_list)} skills after removing overly-generic ones.")

```

```python
# Matching function
def match_skill(skill, desc_words_set, desc_text, core_tokens_map=core_tokens_map):
    """
    Return True if skill matches the description represented by desc_words_set and desc_text.
    Rules:
     - Exact phrase match wins.
     - Else, require at least one core token present.
     - If skill has multiple tokens: require either exact phrase or (>=1 core token AND >=1 other token match)
     - Single-token skills: require that token (it is a core token) present in text words.
    """
    skill_text = skill  # already normalized
    # exact phrase
    if re.search(r"\b" + re.escape(skill_text) + r"\b", desc_text):
        return True

    tokens = skill_tokens_map[skill_text]
    core_tokens = core_tokens_map.get(skill_text, [])

    # if no core tokens, don't match (we removed them earlier but keep check)
    if not core_tokens:
        return False

    # presence of any core token in description
    core_present = any(tok in desc_words_set for tok in core_tokens)
    if not core_present:
        return False

    # if single-token skill (e.g., 'python' or 'react'), core_present is enough
    if len(tokens) == 1:
        return True

    # for multi-token skill: require additional evidence:
    # either the exact phrase (already checked) OR at least one other token (core or non-generic)
    other_tokens = [t for t in tokens if t not in core_tokens] + [t for t in tokens if t in core_tokens]
    # remove duplicates
    other_tokens = list(dict.fromkeys(other_tokens))

    # check if any other token (excluding the matched core token) is present
    # (we'll just check if any token besides the core appears; this is conservative)
    # Actually require at least one token from tokens besides the core_tokens that matched core token.
    other_present = any(tok in desc_words_set for tok in tokens if tok in desc_words_set and tok not in generic_tokens)

    # fallback: if other_present is False, but there are >=2 core tokens and at least 2 core tokens are present -> accept
    core_count_in_desc = sum(1 for tok in core_tokens if tok in desc_words_set)
    if other_present or core_count_in_desc >= 2:
        return True

    return False

def extract_skills_from_text(desc_text):
    desc = clean_text(desc_text)
    words = set(desc.split())
    matches = []
    for skill in safe_skill_list:
        try:
            if match_skill(skill, words, desc):
                matches.append(skill)
        except Exception as e:
            # just in case; won't break pipeline
            continue
    return matches
```
