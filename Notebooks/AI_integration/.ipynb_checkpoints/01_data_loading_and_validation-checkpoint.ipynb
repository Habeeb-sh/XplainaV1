<h3>Load datasets</h3>

```python
courses_path = "../../Data/Final Data/all-courses-hybrid-45-skills.csv"
students_path = "../../Data/Final Data/student_responses.csv"

courses_df = pd.read_csv(courses_path)
students_df = pd.read_csv(students_path)
```

<h3>Identify skill columns</h3>

```python
non_skill_cols = [
    "course_title",
    "rating",
    "review_count",
    "skills_list",
    "platform",
    "level",
    "duration",
    "skills_text"
]

course_skill_cols = [c for c in courses_df.columns if c not in non_skill_cols]

len(course_skill_cols), course_skill_cols
```

<h4>Student skill columns normalization</h4>

```python
students_df_raw = students_df.copy()

def normalize_student_columns(df):
    rename_map = {}
    for col in df.columns:
        if ") " in col:
            skill = col.split(") ", 1)[1].strip().lower()
            rename_map[col] = skill
    return df.rename(columns=rename_map)

students_df = normalize_student_columns(students_df)
students_df.head(2)
```

<h3>Align skill columns</h3>

```python
student_skill_cols = [
    c for c in students_df.columns
    if c not in ["Timestamp"]
]

common_skills = sorted(set(course_skill_cols).intersection(student_skill_cols))

print("Course skills:", len(course_skill_cols))
print("Student skills:", len(student_skill_cols))
print("Common skills:", len(common_skills))
```

<h4>check mismatches</h4>

```python
missing_in_students = set(course_skill_cols) - set(common_skills)
missing_in_courses = set(student_skill_cols) - set(common_skills)

print("Missing in students:", missing_in_students)
print("Missing in courses:", missing_in_courses)
```

<h5>Quick hard check:</h5>

<h2><b>03 Rule Based Scoring</h2>

Goals: By the end of this step, we will:

- Compute hybrid recommendation scores for all courses

- Explicitly separate:

  - weak-skill improvement contribution

  - strong-skill enhancement contribution

- Rank courses
  
- Return Top-5 recommendations

- Visually explain why the top course was chosen

<h3>Locked scoring parameters</h3>

```python
WEAK_MAX = 4
STRONG_MIN = 7
ALPHA = 0.6  # weight for weak-skill improvement
```

<h3>Prepare student vectors</h3>

```python
student_raw = STUDENT_PROFILE["raw"]
student_norm = STUDENT_PROFILE["normalized"]
```

<h4>Create skill masks</h4>

```python
weak_mask = student_raw <= WEAK_MAX
strong_mask = student_raw >= STRONG_MIN
```

These masks ensure:

Weak logic applies ONLY to weak skills

Strong logic applies ONLY to strong skills

<h3>Extract course skill matrix</h3>

```python
course_skill_matrix = courses_clean[SKILL_COLS].values.astype(float)
```

<h4>Shape check</h4>

```python
print("Courses:", course_skill_matrix.shape[0])
print("Skills:", course_skill_matrix.shape[1])
```

<h3>Compute contribution matrices</h3>

<h4>Weak-skill improvement contribution</h4>

Formula per skill: 𝑐𝑜𝑢𝑟𝑠𝑒_ℎ𝑎𝑠_𝑠𝑘𝑖𝑙𝑙×(1−𝑠𝑘𝑖𝑙𝑙/10)

Applied only to weak skills:

```python
weak_gap = (1.0 - student_norm.values) * weak_mask.values
weak_contribution = course_skill_matrix @ weak_gap
```

<h4>Strong-skill enhancement contribution</h4>

Formula per skill: 𝑐𝑜𝑢𝑟𝑠𝑒_ℎ𝑎𝑠_𝑠𝑘𝑖𝑙𝑙×(𝑠𝑘𝑖𝑙𝑙/10)

Applied only to strong skills:

```python
strong_match = student_norm.values * strong_mask.values
strong_contribution = course_skill_matrix @ strong_match
```

<h3>Final hybrid score</h3>

```python
final_score = (
    ALPHA * weak_contribution +
    (1 - ALPHA) * strong_contribution
)
```

<h4>Attach scores:</h4>

```python
courses_scored = courses_clean.copy()
courses_scored["weak_score"] = weak_contribution
courses_scored["strong_score"] = strong_contribution
courses_scored["final_score"] = final_score
```

<h3>Top-5 recommendations</h3>

```python
top_5_courses = courses_scored.sort_values(
    "final_score", ascending=False
).head(5)

top_5_courses[
    ["course_title", "platform", "level", "duration", "rating", "final_score"]
]
```

<h3>Deep explanation of #1 recommendation</h3>

<h4>8.1 Extract top course</h4>

```python
top_course_idx = top_5_courses.index[0]
top_course = courses_scored.loc[top_course_idx]
top_course
```

<h4>Per-skill contribution breakdown</h4>

```python
course_vector = courses_clean.loc[top_course_idx, SKILL_COLS]

weak_skill_contrib = course_vector * weak_gap
strong_skill_contrib = course_vector * strong_match

hybrid_skill_contrib = (
    ALPHA * weak_skill_contrib +
    (1 - ALPHA) * strong_skill_contrib
)

skill_contrib_df = pd.DataFrame({
    "skill": SKILL_COLS,
    "weak_contribution": weak_skill_contrib,
    "strong_contribution": strong_skill_contrib,
    "hybrid_contribution": hybrid_skill_contrib
}).sort_values("hybrid_contribution", ascending=False)
```

<h4>Top contributors:</h4>

```python
skill_contrib_df.head(10)
```

<h3>Visual explanation</h3>

```python
plt.figure(figsize=(10, 4))
skill_contrib_df.head(10).set_index("skill")["hybrid_contribution"].plot(kind="bar")
plt.title("Top Skill Contributions — #1 Recommended Course")
plt.ylabel("Hybrid Contribution Score")
plt.tight_layout()
plt.show()
```

<h3>Sanity validation</h3>

```python
#Top course covers weak skills
#Or reinforces strong ones

covered_weak = [
    s for s in STUDENT_PROFILE["weak_skills"]
    if course_vector[s] == 1
]

covered_strong = [
    s for s in STUDENT_PROFILE["strong_skills"]
    if course_vector[s] == 1
]

covered_weak, covered_strong
```

<h2><b>04 Training Pair Generation (Pseudo-Labels)</h2>

Goals: By the end of this step we will have:

- Converted recommendations into supervised learning data

- Created (student, course) feature vectors

- Generated pseudo-labels

- Produced a clean ML-ready dataset

- Ensured features are SHAP/LIME-friendly

<h3>Extract student data</h3>

```python
student_row = students_clean.loc[STUDENT_INDEX, SKILL_COLS]
student_norm = student_row / 10.0

weak_mask = student_row <= WEAK_MAX
strong_mask = student_row >= STRONG_MIN
```

<h3>Build training pairs (student × course)</h3>

```python
X_rows = []

for idx, course in courses_clean.iterrows():
    row = {}

    for skill in SKILL_COLS:
        course_has_skill = course[skill]
        student_skill_norm = student_norm[skill]

        # Weak gap feature
        row[f"weak_gap_{skill}"] = (
            course_has_skill * (1 - student_skill_norm)
            if student_row[skill] <= WEAK_MAX
            else 0
        )

        # Strong match feature
        row[f"strong_match_{skill}"] = (
            course_has_skill * student_skill_norm
            if student_row[skill] >= STRONG_MIN
            else 0
        )

    X_rows.append(row)

X = pd.DataFrame(X_rows)
X.shape
```

<h3>Generate pseudo-labels</h3>

1 → Top-K courses

0 → Others

<h4>Choose K</h4>

```python
TOP_K = 5
```

<h4>Label creation</h4>

```python
top_k_indices = courses_scored.sort_values(
    "final_score", ascending=False
).head(TOP_K).index

y = courses_scored.index.isin(top_k_indices).astype(int)
```

Sanity check:

```python
np.bincount(y)
```

<h3>Final ML dataset</h3>

```python
X_ml = X.copy()
y_ml = y.copy()

print("X shape:", X_ml.shape)
print("y shape:", y_ml.shape)
```

```python
X_ml.head(2)
```

<h3>Freeze dataset for reuse</h3>

```python
ML_DATASET = {
    "X": X_ml,
    "y": y_ml,
    "feature_names": X_ml.columns.tolist()
}
```

- Train a Logistic Regression model (baseline, interpretable)

<h3>Imports</h3>

```python
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
```

<h3>Extract ML data</h3>

```python
X = ML_DATASET["X"]
y = ML_DATASET["y"]
feature_names = ML_DATASET["feature_names"]

print("X shape:", X.shape)
print("y positives:", y.sum())
```

<h3>Logistic Regression (baseline model)</h3>

Note:

- class_weight="balanced" → handle heavy imbalance

- liblinear → stable for small datasets

```python
log_reg = Pipeline([
    ("scaler", StandardScaler()),
    ("clf", LogisticRegression(
        class_weight="balanced",
        max_iter=1000,
        solver="liblinear"
    ))
])

log_reg.fit(X, y)
```

<h3>Inspect Logistic Regression coefficients</h3>

```python
coef = log_reg.named_steps["clf"].coef_[0]

coef_df = pd.DataFrame({
    "feature": feature_names,
    "coefficient": coef
}).sort_values("coefficient", ascending=False)

coef_df.head(15)
```

```python
coef_df.tail(15)
```

<h3>Logistic Regression ranking sanity check</h3>

```python
logreg_scores = log_reg.predict_proba(X)[:, 1]

courses_scored["logreg_score"] = logreg_scores

courses_scored.sort_values(
    "logreg_score", ascending=False
).head(TOP_K)[
    ["course_title", "final_score", "logreg_score"]
]
```

<h3>Tree-based model (Random Forest)</h3>

```python
rf_model = RandomForestClassifier(
    n_estimators=200,
    max_depth=6,
    class_weight="balanced",
    random_state=42
)

rf_model.fit(X, y)
```

<h3>Tree model ranking sanity check</h3>

```python
rf_scores = rf_model.predict_proba(X)[:, 1]

courses_scored["rf_score"] = rf_scores

courses_scored.sort_values(
    "rf_score", ascending=False
).head(TOP_K)[
    ["course_title", "final_score", "rf_score"]
]
```

<h3>Feature importance (tree model)</h3>

```python
rf_importance = pd.DataFrame({
    "feature": feature_names,
    "importance": rf_model.feature_importances_
}).sort_values("importance", ascending=False)

rf_importance.head(15)
```

<h3>Freeze trained models</h3>

```python
TRAINED_MODELS = {
    "logistic_regression": log_reg,
    "random_forest": rf_model
}
```

<h2><b>06 SHAP Explanations</h2>

Goals: By the end of this step, we will:

- Explain global model behavior (what skills matter overall)

- Explain local decisions (why this course was recommended)

- Use SHAP correctly for each model type

- Produce clear, report-ready plots

- Tie explanations back to student weak/strong skills

<h3>Imports & Extracts</h3>

```python
# !pip install shap
```

```python
import shap
```

```python
X = ML_DATASET["X"]
y = ML_DATASET["y"]
feature_names = ML_DATASET["feature_names"]

log_reg = TRAINED_MODELS["logistic_regression"]
rf_model = TRAINED_MODELS["random_forest"]
```

<h3>SHAP — Logistic Regression (global explanation)</h3>

<h4>Create explainer</h4>

```python
X_scaled = log_reg.named_steps["scaler"].transform(X)
clf = log_reg.named_steps["clf"]

explainer_lr = shap.LinearExplainer(
    clf,
    X_scaled,
    feature_names=feature_names
)
```

<h4>Compute SHAP values</h4>

```python
shap_values_lr = explainer_lr.shap_values(X_scaled)
```

Shape sanity check

```python
np.array(shap_values_lr).shape
```

<h4>Global summary plot (LogReg)</h4>

```python
shap.summary_plot(
    shap_values_lr,
    X_scaled,
    feature_names=feature_names,
    show=True
)
```

<h3>SHAP — Tree model (Random Forest)</h3>

<h4>Create TreeExplainer</h4>

```python
explainer_rf = shap.TreeExplainer(rf_model)
```

<h4>Compute SHAP values</h4>

```python
shap_values_rf = explainer_rf.shap_values(X)
print("type:", type(shap_values_rf))
print("shape:", np.array(shap_values_rf).shape)
```

```python
shap_values_rf_pos = shap_values_rf[:, :, 1]

print("X shape:", X.shape)
print("shap_values_rf_pos shape:", shap_values_rf_pos.shape)
```

<h4>Global summary plot (Tree model)</h4>

```python
shap.summary_plot(
    shap_values_rf_pos,
    X,
    feature_names=feature_names,
    show=True
)
```

<h3>Local SHAP explanation — single recommendation</h3>

<h4>Pick top recommended course</h4>

```python
top_course_idx = courses_scored.sort_values(
    "rf_score", ascending=False
).index[0]

top_course_idx
```

```python
#Extract its feature row:
x_instance = X.loc[top_course_idx]
```

<h4>SHAP waterfall plot (tree model)</h4>

```python
shap.waterfall_plot(
    shap.Explanation(
        values=shap_values_rf_pos[top_course_idx],
        base_values=explainer_rf.expected_value[1],
        data=x_instance,
        feature_names=feature_names
    )
)
```

This plot explains:

Which skills pushed the decision up

Which skills pushed it down

Final recommendation probability

<h3>Mapping SHAP back to skills (human explanation)</h3>

```python
shap_contrib = pd.DataFrame({
    "feature": feature_names,
    "shap_value": shap_values_rf_pos[top_course_idx]
}).sort_values("shap_value", ascending=False)

shap_contrib.head(10)
```

<h4>convert to natural language</h4>

```python
def explain_feature(feature):
    if feature.startswith("weak_gap_"):
        return f"Improves weak skill: {feature.replace('weak_gap_', '')}"
    if feature.startswith("strong_match_"):
        return f"Enhances strong skill: {feature.replace('strong_match_', '')}"
    return feature

shap_contrib["explanation"] = shap_contrib["feature"].apply(explain_feature)

shap_contrib.head(10)[["feature", "shap_value", "explanation"]]
```

<h3>global importance bar plot</h3>

```python
# Global importance: mean(|SHAP|)
global_importance = np.abs(shap_values_rf_pos).mean(axis=0)

imp_df = pd.DataFrame({
    "feature": feature_names,
    "mean_abs_shap": global_importance
}).sort_values("mean_abs_shap", ascending=False).head(15)

plt.figure(figsize=(10,4))
plt.bar(imp_df["feature"][::-1], imp_df["mean_abs_shap"][::-1])
plt.title("Global Feature Importance (Random Forest) — mean(|SHAP|)")
plt.xlabel("mean(|SHAP|)")
plt.tight_layout()
plt.show()

imp_df
```

<h3>Explain Top-5 recommendations, not only #1</h3>

```python
TOP_K = 5
top_labels = courses_scored.sort_values("rf_score", ascending=False).head(TOP_K).index.tolist()

explanations_top5 = []

for lbl in top_labels:
    pos = X.index.get_loc(lbl)
    contrib = pd.DataFrame({
        "feature": feature_names,
        "shap_value": shap_values_rf_pos[pos]
    }).sort_values("shap_value", ascending=False).head(6)

    contrib["reason"] = contrib["feature"].apply(explain_feature)

    explanations_top5.append({
        "course_title": courses_scored.loc[lbl, "course_title"],
        "top_reasons": "; ".join([f"{r} ({v:.3f})" for r, v in zip(contrib["reason"], contrib["shap_value"])])
    })

pd.DataFrame(explanations_top5)
```

<h2><b>07 LIME Explanations</h2>

Goals: By the end of this step, we will:

- Generate a local explanation for the top recommended course

- Show the top contributing features

- Convert features into human-readable reasons

- Do it for Top-5

<h3>Install</h3>

```python
# !pip install lime
from lime.lime_tabular import LimeTabularExplainer
```

<h3>Build the LIME explainer</h3>

```python
explainer_lime = LimeTabularExplainer(
    training_data=X.values,
    feature_names=feature_names,
    class_names=["not_recommended", "recommended"],
    mode="classification",
    discretize_continuous=True  # good default for tabular
)
```

<h3>Choose the instance (top recommended course)</h3>

```python
top_course_label = courses_scored.sort_values("rf_score", ascending=False).index[0]
top_course_pos = X.index.get_loc(top_course_label)

x_instance = X.loc[top_course_label].values

print("Top course title:", courses_scored.loc[top_course_label, "course_title"])
```

<h3>Run LIME explanation</h3>

```python
lime_exp = explainer_lime.explain_instance(
    data_row=x_instance,
    predict_fn=rf_model.predict_proba,
    num_features=12  # show top 12
)
```

<h4>Visual</h4>

```python
lime_exp.show_in_notebook(show_table=True)
```

<h4>Table explanation</h4>

```python
lime_list = lime_exp.as_list()  # list of (feature_rule, weight)

lime_df = pd.DataFrame(lime_list, columns=["lime_rule", "weight"]).sort_values("weight", ascending=False)
lime_df
```

<h3>Convert LIME “rules” into skill-friendly reasons</h3>

```python
import re

def extract_feature_name(lime_rule: str) -> str:
    # takes "feature <= value" or "feature > value" etc.
    return re.split(r"\s<=\s|\s>=\s|\s<\s|\s>\s|\s=\s", lime_rule)[0].strip()

lime_df["feature"] = lime_df["lime_rule"].apply(extract_feature_name)
lime_df["reason"] = lime_df["feature"].apply(explain_feature)

lime_df[["feature", "weight", "reason", "lime_rule"]].head(12)
```

<h3>Plot LIME weights</h3>

```python
plot_df = lime_df.head(12).copy()

plt.figure(figsize=(10,4))
plt.bar(plot_df["reason"][::-1], plot_df["weight"][::-1])
plt.title("LIME Explanation — #1 Recommended Course")
plt.ylabel("LIME weight (impact on recommendation)")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

<h3>Explain Top-5 recommendations, not only #1</h3>

```python
TOP_K = 5
top_labels = courses_scored.sort_values("rf_score", ascending=False).head(TOP_K).index.tolist()

lime_summaries = []

for lbl in top_labels:
    pos = X.index.get_loc(lbl)
    x_row = X.loc[lbl].values

    exp = explainer_lime.explain_instance(
        data_row=x_row,
        predict_fn=rf_model.predict_proba,
        num_features=6
    )

    df = pd.DataFrame(exp.as_list(), columns=["lime_rule", "weight"]).sort_values("weight", ascending=False)
    df["feature"] = df["lime_rule"].apply(extract_feature_name)
    df["reason"] = df["feature"].apply(explain_feature)

    lime_summaries.append({
        "course_title": courses_scored.loc[lbl, "course_title"],
        "top_reasons": "; ".join([f"{r} ({w:.3f})" for r, w in zip(df["reason"], df["weight"])])
    })

pd.DataFrame(lime_summaries)
```
